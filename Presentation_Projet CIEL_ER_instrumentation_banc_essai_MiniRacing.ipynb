## I. Contexte du projet

<center><a href="https://mr79niort.wordpress.com/infos-pratiques/"><img title="" src="imgs/presentation.png" width=70%></a></center>

<div style="text-align: justify">Créé en 1983 à Niort, le Mini Racing 79 est une association à but non lucratif ayant pour vocation la pratique du modélisme automobile radiocommandé. Par son affiliation à la fédération de Voiture Radio-Commandée (FVRC), le Mini Racing 79 permet à ses adhérents de participer à des courses dans toute la France.

<br/>
Le club Mini Racing 79 a pour projet d’améliorer l’instrumentation de son banc d’essai pour mesurer et afficher certaines caractéristiques des voitures et ainsi permettre la réalisation de réglages des performances des différents types de voitures RC (1/8 électrique, 1/8 thermique, 1/10 TT). Le club souhaite également équiper les voitures RC et le banc d’essais d’un système de transpondeur RFID pour identifier chaque voiture et implémenter un suivi de ses performances et réglages.</div>

---

<center><img title="" src="imgs/bancEssais.JPEG" width=60%></center>

<div style="text-align: justify">Le banc d'essai existant permet de tester et d'évaluer divers aspects des performances d'un modèle réduit de voiture RC afin d'optimiser les réglages et d'assurer une performance maximale avant les essais sur le terrain. 
Un banc d'essai permet notamment de mesurer et d'optimiser les performances moteur, d'affiner le freinage et l'accélération, d'optimiser la batterie.</div>

###  Caractéristiques du banc
* Adaptabilité aux différents types de voitures :
    * Electrique
    * Thermique
    * Echelle 1/8 ou 1/10
* Mesures et essais :
    * Couple moteur
    * Vitesse de rotation des roues
    * Accélération
    * Cycle de fatigue
    * Consommation de la voiture

## II.Expression fonctionnelle du besoin (ce que le client veut)

Le club Mini Racing souhaite instrumenter le banc d'essai de manière à réaliser des mesures automatiques et un suivi des performances des voitures. 

**Mesures et affichage :**
* Mesure de la vitesse de rotation des roues
* Mesure ou calcul du couple moteur
* Mesure du courant de la batterie
* Mesure de la tension de la batterie
* Calcul du taux de charge de la batterie
* Affichage des performances de la voiture sur écran local
  

**Implémentation du suivi des performances des voitures :**
* Installation d’un système de transpondeur RFID sur les voitures
* Installation d’un système de lecture sur le banc pour l’identification de chaque voiture
* Enregistrement de l'identifiant et des caractéristiques de la voiture dans une base de données pour un suivi des différents essais
* Implémentation d'un serveur web pour un affichage déporté des performances
   

**Connectivité :**
* Liaison filaire entre la carte d'acquisition et l'écran local
* Liaison sans fil pour l'affichage déporté des performances
  
**Autonomie et alimentation :**
* Alimentation sur secteur du banc
* Alimentation sur batterie des voitures
