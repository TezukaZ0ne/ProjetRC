# Côté Schéma électronique

Pour créer des BUS:
* placer un BUS (clic droit Plcer ==> BUS)
* Réaliser vos liens
* Ajouter un label de fil (clic droit sur le fil) ==> même label réalise la connexion

# Côté PCB (typon)

* Créer la surface de votre carte ==> lien avec la couche "Board Edge" (jaune)
* placer vos composants
* Placer des zone KeepOut pour interdir la connexion des pistes coté composant:
    * sélectionner la couche "keepout"
    * sélectionner Rectangle Graphique 2D
    * tracer le rectangle voulu
    * éditer le rectangle pour changer la couche ==> Top Copper (côté composants)
* Placer du texte Côté Composant et Côté Cuivre
